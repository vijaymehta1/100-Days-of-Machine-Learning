```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt  
import seaborn as sns 
```

```python
from sklearn.preprocessing import KBinsDiscretizer
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('train.csv',usecols=['Survived','Fare','Age'])
df.shape
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
df.shape
```

```python
x = df.iloc[:,1:]
y = df.iloc[:,0]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,random_state=42,test_size=.2)
```

```python
dt = DecisionTreeClassifier()
dt.fit(x_train,y_train)
y_pred =dt.predict(x_test)
accuracy_score(y_test,y_pred)
```

```python
np.mean(cross_val_score(DecisionTreeClassifier(),x,y,cv=10,scoring='accuracy'))
```

```python
bin_age =KBinsDiscretizer(n_bins=15,encode='ordinal',quantile_method='averaged_inverted_cdf')
bin_fare=KBinsDiscretizer(n_bins=15 ,encode="ordinal",quantile_method='averaged_inverted_cdf')
```

```python
trf = ColumnTransformer(
    [
        ('bin_age',bin_age,[0]),
        ('bin_fare',bin_fare,[1])
    ]
)
```

```python
x_train_trans=trf.fit_transform(x_train)
y_test_trans = trf.transform(x_test)
```

```python
clf = DecisionTreeClassifier()

clf.fit(x_train_trans,y_train)
pred1 = clf.predict(x_test)

accuracy_score(y_test,pred1)
```
