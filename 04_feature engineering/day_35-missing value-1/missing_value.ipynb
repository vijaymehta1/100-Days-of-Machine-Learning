```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df  = pd.read_csv("data_science_job (10).csv")
```

```python
df.shape
```

```python
df.isnull().mean()*100
```

```python
cols = [ var for var in df.columns if df[var].isnull().mean()<.05 and df[var].isnull().mean()>0]
cols
```

```python
df['education_level'].value_counts()
```

```python
len(df[cols].dropna()) / len(df)
```

```python
new_df = df[cols].dropna()
df.shape,new_df.shape
```

```python
new_df.hist(bins=50,figsize=(10,10))
```

```python
plt.Figure(figsize=(12,8))
sns.histplot(df['training_hours'],bins=50,color='red',stat='density',label= 'original data')
sns.histplot(new_df['training_hours'],bins=50,stat='density',color='green',label='after cca')
```

```python
sns.kdeplot(df['training_hours'],color='green')
sns.kdeplot(new_df['training_hours'],color='red')
```

```python
plt.Figure(figsize=(12,8))
sns.histplot(df['experience'],bins=50,color='red',stat='density',label= 'original data')
sns.histplot(new_df['experience'],bins=50,stat='density',color='green',label='after cca')
```

```python
sns.kdeplot(df['experience'],color='green')
sns.kdeplot(new_df['experience'],color='red')
```

```python
plt.Figure(figsize=(12,8))
sns.histplot(df['city_development_index'],bins=50,color='red',stat='density',label= 'original data')
sns.histplot(new_df['city_development_index'],bins=50,stat='density',color='green',label='after cca')
```

```python
sns.kdeplot(df['city_development_index'],color='green')
sns.kdeplot(new_df['city_development_index'],color='red')
```
